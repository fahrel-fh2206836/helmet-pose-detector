```python
import timm
import torch

model = timm.create_model("convnext_tiny", pretrained=True, num_classes=2)

num_params = sum(p.numel() for p in model.parameters())
print(f"Total params: {num_params:,}")
print(f"Total params (M): {num_params / 1e6:.2f}M")
```
